# cleaning the unnecessary ratings are occured in ratings

```python


valid_ratings=['Parents Strongly Cautioned', 'TV-Mature Audience Only',
       'Parental Guidance Suggested', 'TV-May be inapproproate under 14',
       'Tv-Parental Guidance', 'TV-Y', 'TV-Y7', 'Restricted',
       'TV- General Audience', 'General Audience', 'Adults Only', 'NR', 'TV-Y7-FV', 'UR']
```

```python
netflix_df["rating"]=netflix_df["rating"].apply(lambda x:x if x in valid_ratings else np.nan)
```

```python
netflix_df["rating"].fillna("Not Rated")
```

```python
netflix_df.dropna(subset=["rating"],inplace=True)
```

## Bivariate/ Multivariate Analysis

```python
netflix1_df=netflix_df.groupby(["type","release_year","title","country"]).size().reset_index().rename(columns={0:"count"})
netflix1_df
```

```python
sns.histplot(data=netflix1_df,x="release_year",bins=30,kde=True,color="black",line_kws={"color":"green","linewidth":2})
plt.title("distribution of netflix tiltes by release year")
plt.xlabel("release Year")
plt.ylabel("number of tiltes")
plt.tight_layout()
```

## Observation:-

 Very less number of tiltes before 1980
 
There is titles increasing after 2000.
The highest titles shows on 2020 above 3500 titles
